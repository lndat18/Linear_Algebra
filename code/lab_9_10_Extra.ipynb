```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Ex12
# y = a0 + a1x

def draw_2D(a0, a1, data_x, data_y, estimate_x):
  x = np.linspace(2000, 2030, 100)
  y = a0 + a1*x
  estimate_y = a0 + a1*estimate_x
  print(f"estimate the sales in the year 2020: {estimate_y:.2f}")
  plt.plot(x, y, label=f"y = {a0:.2f} + {a1:.2f}x")
  plt.scatter(data_x, data_y, color='hotpink', label="data point")
  plt.scatter(estimate_x, estimate_y, color="red", label="estimate")
  plt.xlabel("Years")
  plt.ylabel("Scales")
  plt.legend()
  plt.grid()
  plt.show()

x = np.array([2015,2016,2017,2018,2019])
y = np.array([12,19,29,37,45])
x_add = np.vstack((x, np.ones(len(x)))).T

a1, a0 = np.linalg.lstsq(x_add, y, rcond=None)[0]

estimate_x = 2020

draw_2D(a0, a1, x, y, estimate_x)
```

```python
A = np.array([[1,2],
              [0,-3],
              [2,6]])

b = np.array([1,1,0])

x,y = np.linalg.lstsq(A, b, rcond=None)[0]

closest_vector = x * np.array([1, 0, 2]) + y * np.array([2, -3, 6])

print(f"The closest vector in the column space of A is: {closest_vector}")
```
