```python
# Ex18
first_row = A[0,:]
last_row = A[-1,:]
first_column = A[:, 0]
last_column = A[:, -1]

sum = np.sum(np.concatenate((first_row, last_row, first_column, last_column)))

sum
```

```python
# Ex19
n = A.shape[0]

# Calculate sum of major and minor diagonals
major_diag_sum = np.sum(np.diag(A))  # Sum of major diagonal
minor_diag_sum = np.sum(np.diag(np.fliplr(A)))  # Sum of minor diagonal

# Handle the case where the center element is counted twice (only for odd n)
if n % 2 != 0:
    center_value = A[n // 2, n // 2]
    total_sum = major_diag_sum + minor_diag_sum - center_value
else:
    total_sum = major_diag_sum + minor_diag_sum

print("Sum of major and minor diagonal elements:", total_sum)
```
