```python
# Ex18

def magic_square(n):
    if n % 2 == 0:
        raise ValueError("Magic square is only possible for odd dimensions.")

    magic_square = np.zeros((n, n), dtype=int)

    num = 1
    i, j = 0, n // 2

    while num <= n**2:
        magic_square[i, j] = num
        num += 1
        new_i, new_j = (i - 1) % n, (j + 1) % n

        if magic_square[new_i, new_j]:
            i += 1
        else:
            i, j = new_i, new_j

    return magic_square

n = 15
magic_square_matrix = magic_square(n)

magic_square_matrix
```

```python
# Ex22
matrix = np.random.randint(1, 101, size=(100, 100))

multiples_of_3_or_5 = (matrix % 3 == 0) | (matrix % 5 == 0)

total_sum = np.sum(matrix[multiples_of_3_or_5])

print("Sum of all multiples of 3 or 5:", total_sum)
```
