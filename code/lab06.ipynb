```python
# Ex1
def norm_1(A):
  return max(sum(abs(A[:, j])) for j in range(A.shape[1]))

A1 = np.array([[1,-7],
              [-2,-3]])

A2 = np.array([[-2,8],
               [3,1]])

A3 = np.array([[2,-8],
               [3,1]])

A4 = np.array([[2,3],
               [1,-1]])

A5 = np.array([[5,-4,2],
               [-1,2,3],
               [-2,1,0]])

print("norm 1 by command")
print(norm_1(A1))
print(norm_1(A2))
print(norm_1(A3))
print(norm_1(A4))
print(norm_1(A5))

# verify the result
print("norm 1 by built in function")
print(np.linalg.norm(A1, ord=1))
print(np.linalg.norm(A2, ord=1))
print(np.linalg.norm(A3, ord=1))
print(np.linalg.norm(A4, ord=1))
print(np.linalg.norm(A5, ord=1))
```

```python
# Ex2
def norm_inf(B):
  return max(sum(abs(B[i, :])) for i in range(B.shape[0]))

B1 = np.array([[1,-7],
               [-2,-3]])

B2 = np.array([[3,6],
               [1,0]])

B3 = np.array([[5,-4,2],
               [-1,2,3],
               [-2,1,0]])

B4 = np.array([[3,6,-1],
               [3,1,0],
               [2,4,-7]])

B5 = np.array([[-3,0,0],
               [0,4,0],
               [0,0,2]])

print("norm inf by command")
print(norm_inf(B1))
print(norm_inf(B2))
print(norm_inf(B3))
print(norm_inf(B4))
print(norm_inf(B5))
```

```python
# Ex 3
def norm_frobenius(C):
  return np.sqrt(np.sum(np.square(C)))

C1 = np.array([[5,-4,2],
               [-1,2,3],
               [-2,1,0]])

C2 = np.array([[1,7,3],
               [4,-2,-2],
               [-2,-1,1]])

C3 = np.array([[2,3],
               [1,-1]])

print("norm Frobenius by command")
print(norm_frobenius(C1))
print(norm_frobenius(C2))
print(norm_frobenius(C3))
```

```python
# Ex4
def angel(u,v):
  dot_product = np.dot(u,v)

  norm_u = np.linalg.norm(u)
  norm_v = np.linalg.norm(v)

  if norm_u == 0 or norm_v == 0:
    raise ValueError("One of the vectors has zero magnitude, angle is undefined.")

  cos_theta = dot_product/(norm_u * norm_v)
  res = np.arccos(cos_theta)

  return np.degrees(res)

# a
u = np.array([1,1])
v = np.array([0,1])
print(angel(u,v), end="\n--------------\n")

# b
u = np.array([1,0])
v = np.array([0,1])
print(angel(u,v), end="\n--------------\n")

# c
u = np.array([-2,3])
v = np.array([1/2,-1/2])
print(angel(u,v), end="\n--------------\n")
```

```python
# Ex5
def unit_vector(u):
  norm_u = np.linalg.norm(u)
  return u/norm_u

# a
u = np.array([2,3])
print(unit_vector(u), end="\n------------------------\n")

# b
u = np.array([1,2,3])
print(unit_vector(u), end="\n------------------------\n")

# c
u = np.array([1/2,-1/2,1/4])
print(unit_vector(u), end="\n------------------------\n")

# d
u = np.array([m.sqrt(2),2,-m.sqrt(2),m.sqrt(2)])
print(unit_vector(u), end="\n------------------------\n")
```

```python
# Ex6
def euclidean_distance(v1, v2):
  return np.linalg.norm((v1-v2))

v1 = np.array([1,2,3])
v2 = np.array([7,4,3])
v3 = np.array([2,1,9])

d_v1_v2 = euclidean_distance(v1,v2)
d_v1_v3 = euclidean_distance(v1,v3)
d_v2_v3 = euclidean_distance(v2,v3)

print(f"Distance between v1 and s2: {d_v1_v2}")
print(f"Distance between v1 and s3: {d_v1_v3}")
print(f"Distance between s2 and s3: {d_v2_v3}")
```

```python
# Ex7
def decode(E, A):
  A_inv = np.linalg.inv(A)
  D = A_inv @ E
  D = np.round(D).astype(int)
  print(D)
  temp = []
  for col in range(D.shape[1]):
    for row in range(D.shape[0]):
      if D[row,col] == 30:
        print("")
      else:
        temp.append(chr(D[row,col] + 61))
  message = ''.join(temp)
  return message

E = np.array([[80,98,99,85,106,94],
              [71,92,76,95,100,92],
              [124,163,140,160,176,161]])

A = np.array([[1,2,3],
              [2,1,2],
              [3,2,4]])

decoded_message = decode(E, A)
print("Decoded Message:", decoded_message)
```

```python
# Ex8
space_value = 100

def encode_message(message, A):
    message_numbers = [ord(char)-61 if char != " " else space_value for char in message]

    while len(message_numbers) % 3 != 0:
        message_numbers.append(space_value)
    E = np.array(message_numbers).reshape(3, -1)
    encoded = A @ E
    return encoded

A = np.array([
    [3, 4, 5],
    [1, 3, 1],
    [1, 1, 2]
])

message1 = "ATTACK"
message2 = "LINEAR ALGEBRA LABORATORY"

encoded1 = encode_message(message1, A)
encoded2 = encode_message(message2, A)

print("Encoded Message 1:\n", encoded1)
print("Encoded Message 2:\n", encoded2)
```

```python
# Ex9 by student
def cosine_similarity_matrix(doc_matrix, q=None):
  norms = np.linalg.norm(doc_matrix, axis=1)

  if q is None:
    row, col = doc_matrix.shape[0], doc_matrix.shape[0]
    similarity_matrix = np.zeros((row, col))
  else:
    row, col = doc_matrix.shape[0], 1
    similarity_matrix = np.zeros((row, col))

  for i in range(row):
    for j in range(col):
      comparison_vector = q if q is not None else doc_matrix[j]
      comparison_norm = np.linalg.norm(q) if q is not None else norms[j]

      dot_product = np.dot(doc_matrix[i], comparison_vector)
      norm_i_j = norms[i] * comparison_norm
      similarity_matrix[i,j] = dot_product/norm_i_j

  return similarity_matrix

doc_matrix = np.array([[0,4,0,0,0,2,1,3],
                      [3,1,4,3,1,2,0,1],
                      [3,0,0,0,3,0,3,0],
                      [0,1,0,3,0,0,2,0],
                      [2,2,2,3,1,4,0,2]])

res = cosine_similarity_matrix(doc_matrix)
print(res)
```
