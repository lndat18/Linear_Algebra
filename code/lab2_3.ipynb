```python
# Exercise 10
A = np.array([[-1,4,8],
              [-9,1,2]])

B = np.array([[5,8],
              [0,-6],
              [5,6]])

C = np.array([[-4,1],
              [6,5]])

D = np.array([[-6,3,1],
              [8,9,-2],
              [6,-1,5]])

def Multiply(m1, m2):
  row1, col1 = m1.shape
  row2, col2 = m2.shape
  if col1 == row2:
    return m1 @ m2
  else:
    return 'Dimensional error: Can not multiply'

def Add(m1, m2):
  if m1.shape == m2.shape:
    return m1 + m2
  else:
    return 'Dimensional error: Can not Add'

def Sub(m1, m2):
  if m1.shape == m2.shape:
    return m1 - m2
  else:
    return 'Dimensional error: Can not Sub'

a = Multiply(A, B.T)
b = Multiply(B, C.T)
c = Sub(C, C.T)
d = Sub(D, D.T)
e = D.T.T
f = 2*C.T
g = A.T + B
h = Add(A.T, B).T
i = Sub(2*A.T, 5*B).T
j = (-D).T
k = (-D).T
l = (C**2).T
m = (C.T)**2

print(a,b,c,d,e,f,g,h,i,j,k,l,m, sep= '\n' + '-'*50 + '\n')
```

```python
# Exercise 11
A = np.array([[2,4,1],
              [6,7,2],
              [3,5,9]])

def isSquare(A):
  return True if A.shape[0] == A.shape[1] else False

def isSymmetric(A):
  return True if np.array_equal(A, A.T) else False

def isSkewSymmetric(A):
  return True if np.array_equal(-A, A.T) else False

a = isSquare(A)
b = isSymmetric(A)
c = isSkewSymmetric(A)
d = np.triu(A)
e = np.tril(A)

print(a,b,c,d,e, sep= '\n' + '-'*50 + '\n')
```

```python
# Exercise 12
A = np.array([[6,0,0,5],
              [1,7,2,-5],
              [2,0,0,0],
              [8,3,1,8]])

B = np.array([[1,-2,5,2],
              [0,0,3,0],
              [2,-6,-7,5],
              [5,0,4,4]])

C = np.array([[3,5,-8,4],
              [0,-2,3,-7],
              [0,0,1,5],
              [0,0,0,2]])

D = np.array([[4,0,0,0],
              [7,-1,0,0],
              [2,6,3,0],
              [5,-8,3,0],
              [5,-8,4,-3]])

E = np.array([[4,0,-7,3,-5],
              [0,0,2,0,0],
              [7,3,-6,4,-8],
              [5,0,5,2,-3],
              [0,0,9,-1,2]])

F = np.array([[6,3,2,4,0],
              [9,0,-4,1,0],
              [8,-5,6,7,1],
              [3,0,0,0,0],
              [4,2,3,2,0]])


def isSquare(matrix):
  row,col = matrix.shape
  return True if row == col else False

def det(matrix):
  return np.linalg.det(matrix) if isSquare(matrix) else 'matrix must be square'


det_A = det(A)
det_B = det(B)
det_C = det(C)
det_D = det(D)
det_E = det(E)
det_F = det(F)

print(det_A, det_B, det_C, det_D, det_E, det_F, sep= '\n' + '-'*50 + '\n')
```

```python
# Exercise 13

def determinant_difference(n):
    A = np.random.randint(0,10, size=(n, n))
    B = np.random.randint(0,10, size=(n, n))

    det_A = np.linalg.det(A)
    det_B = np.linalg.det(B)
    det_A_plus_B = np.linalg.det(A + B)

    difference = det_A_plus_B - det_A - det_B
    return difference

sizes = [2, 3, 4, 5]

determinant_differences = {n: determinant_difference(n) for n in sizes}
determinant_differences
```

```python
# Exercise 14

def determinant_product_check(n):
    A = np.random.randint(0,10, size=(n, n))
    B = np.random.randint(0,10, size=(n, n))

    det_A = np.linalg.det(A)
    det_B = np.linalg.det(B)
    det_AB = np.linalg.det(np.dot(A, B))

    product_det_A_B = det_A * det_B
    return det_AB, product_det_A_B

sizes = [2, 3, 4, 5]

determinant_product_results = [determinant_product_check(n) for n in sizes]
determinant_product_results
```
